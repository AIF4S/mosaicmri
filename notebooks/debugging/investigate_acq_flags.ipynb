```python
import ismrmrd
filename = "meas_MID00018_FID110986_COR_PD_FS.h5"
path = f"/data/datasets/msk_mri_debug/h5/{filename}"
 
dset = ismrmrd.Dataset(path, 'dataset', create_if_needed=False)
 
n_acq = dset.number_of_acquisitions()
for i in range(n_acq):
    item = dset.read_acquisition(i)
    if isinstance(item, ismrmrd.Acquisition):
        if item.is_flag_set(ismrmrd.ACQ_IS_NOISE_MEASUREMENT):
            print(i, "NOISE_MEASUREMENT")
        elif item.is_flag_set(ismrmrd.ACQ_IS_PARALLEL_CALIBRATION):
            print(i, "PARALLEL_CALIBRATION")
        elif item.is_flag_set(ismrmrd.ACQ_IS_PHASECORR_DATA):
            print(i, "PHASECORR")
        elif item.is_flag_set(ismrmrd.ACQ_IS_NAVIGATION_DATA):
            print(i, "NAVIGATION")
        else:
            continue
dset.close()
```
